```python
import os, joblib, numpy as np, torch
import torch.nn.functional as f
from torchvision import datasets, transforms

import joblib
import os
```

# IMAGE PREPROCESS

```python
name, DS, mean, n_val = "mnist", datasets.MNIST, (0.5,), 10000
tf = transforms.Compose([transforms.ToTensor(), transforms.Normalize(mean, mean)])

def to_tensors(ds):
    X = torch.stack([x for x, _ in ds])
    y = f.one_hot(torch.as_tensor(ds.targets), 10).float()
    return X, y

X, y = to_tensors(DS("datasets/_raw", train=True, download=True, transform=tf))
X_test, y_test = to_tensors(DS("datasets/_raw", train=False, download=True, transform=tf))
splits = {"train": (X[:-n_val], y[:-n_val]), "val": (X[-n_val:], y[-n_val:]), "test": (X_test, y_test)}

folder = os.path.join("datasets", name)
os.makedirs(folder, exist_ok=True)
for s, (Xs, ys) in splits.items():
    np.save(os.path.join(folder, f"X{s}_or"), Xs.numpy())
    torch.save(Xs, os.path.join(folder, f"X{s}.pt"))
    torch.save(ys, os.path.join(folder, f"y{s}.pt"))

joblib.dump(dict(feature_names=None, categorical_features=[], categorical_names=None, ordinal=[],
                 discrete=[], encoder=None, num_features=None),
            os.path.join(folder, "info_dict.joblib"))

info_path = os.path.join("load", "dataset_info.joblib")
info = joblib.load(info_path)
info[name] = dict(file_name=None, separator=None, target_name=None, remove_list=[], categorical_features=[],
                  ordinal=[], discrete=[], train_size=None, test_size=None)
joblib.dump(info, info_path)
```
